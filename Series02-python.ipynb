# IEAP-python-series02: Sinusoids, from math to code

In this prject, we will plot how to plot sinus waves using python
The maths involved with this are

$y(t) = A \sin(\omega t + \varphi)$

Where:
- $y(t)$ is the signal value at time $t$
- $A$ is the amplitude of the sine wave
- $\omega$ is the angular frequency
- $t$ is the time variable, here the array created with `np.linspace(0, 1, 1000)`
- $\varphi$ is the phase shift

```python
# Import the needed libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time). So that we have a time column to work with
t = np.linspace(0, 1, 1000)

# Create a numpy array called s that contains the values of a sine function of t (s is for signal). So that we have sine data. 
s = np.sin(2 * np.pi * 1 * t)

# Plotting s as a function of t
plt.plot(t, s)

# Adding a title to the plot
plt.title('Sine Wave')

# Adding a label to the x axis
plt.xlabel('Time (s)')

# Adding a label to the y axis
plt.ylabel('Amplitude')

# Adding a grid to the plot
plt.grid(True)

# Saving the plot as a png file
plt.savefig('sine_wave.png')
```

```python


# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)

# Create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = np.sin(2 * np.pi * 1 * t)

# Plot s as a function of t
plt.plot(t, s, label='Sinus')

# Add a title to the plot
plt.title('Sine Wave')

# Add a label to the x axis
plt.xlabel('Time (s)')

# Add a label to the y axis
plt.ylabel('Amplitude')
# Add a grid to the plot
plt.grid(True)
# Save the plot as a png file
plt.savefig('sine_wave_and_cosine_wave.png')

#adding the cosine data to the plot(the cosine data is created within the function to improve coding)
plt.plot(t, np.cos(2 * np.pi * 1 * t), label='Cosinus')

# Add a legend to the plot
plt.legend(loc='lower left')

# get the current axis(gca) and add a second(twinx) y axis to the plot(both are in the same code)
ax2 = plt.gca().twinx()

# Add a second label to the y axis
ax2.set_ylabel('Amplitude')

# Save the plot as a pdf file  
plt.savefig('sine_and_cosine_wave.pdf')
```
