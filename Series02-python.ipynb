# IEAP-python-series02: Sinusoids, from math to code

In this prject, we will plot how to plot sinus waves using python
The maths involved with this are

$y(t) = A \sin(\omega t + \varphi)$

Where:
- $y(t)$ is the signal value at time $t$
- $A$ is the amplitude of the sine wave
- $\omega$ is the angular frequency
- $t$ is the time variable, here the array created with `np.linspace(0, 1, 1000)`
- $\varphi$ is the phase shift

```python
# Import the needed libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)

# Create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = np.sin(2 * np.pi * 1 * t)

# Plot s as a function of t
plt.plot(t, s)

# Add a title to the plot
plt.title('Sine Wave')

# Add a label to the x axis
plt.xlabel('Time (s)')

# Add a label to the y axis
plt.ylabel('Amplitude')
# Add a grid to the plot
plt.grid(True)
# Save the plot as a png file
plt.savefig('sine_wave.png')
```
